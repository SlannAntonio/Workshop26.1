```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt


#Numpy temperaturas e dias
temperaturas = np.random.randint(20,36,7)
diasem = ['Domingo','Segunda-feira','Terça-feira','Quarta-feria','Quimta-feira','Sexta-feira','Sábado']
for dia, temp in  zip(diasem,temperaturas):
  print(f"{dia}:{temp}°C")
```

```python
#Pandas DF dias e suas temps
df = pd.DataFrame({
    'Dia':diasem,
    'Temperatura °C': temperaturas
})

df

```

```python
#Média das temps
medtemp = df['Temperatura °C'].mean()
print(f'A média da temperatura na semana foi de {medtemp}°C\n')

#Temperatura max e min
maxtemp = df['Temperatura °C'].max()
mintemp = df['Temperatura °C'].min()

print(f'O máximo de temperatura alcançada esta semana foi de {maxtemp}°C')
print(f'O mínimo de temperatura alcançada esta semana foi de {mintemp}°C')

```

```python
#Primeira Tabela
df.plot(x ='Dia', y='Temperatura °C', kind = 'line')
plt.xticks(rotation =45)
plt.show()
```

```python
#Segunda TAbela
df.plot(x ='Dia', y = 'Temperatura °C', kind ='bar')
plt.xticks(rotation = 45)
plt.show()
```

```python
#terceira Tabela
df.plot( x = 'Dia', y = 'Temperatura °C', kind = 'area')
plt.xticks(rotation = 45)
plt.show()
```
